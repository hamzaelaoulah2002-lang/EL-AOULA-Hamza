```python
import math
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# ==========================================================
# 1. DONNÉES INDICATIVES
# ==========================================================

SECTIONS = [1.5, 2.5, 4, 6, 10, 16, 25, 35, 50, 70, 95, 120, 150, 185, 240]

# Intensités admissibles indicatives (A).
# À remplacer par les valeurs normatives correspondant au mode
# de pose, au type d'isolant et aux conditions réelles.
IZ_CU = {
    1.5: 16, 2.5: 21, 4: 28, 6: 36, 10: 50,
    16: 68, 25: 89, 35: 110, 50: 134, 70: 171,
    95: 207, 120: 239, 150: 275, 185: 314, 240: 370
}

IZ_AL = {
    1.5: 0, 2.5: 0, 4: 22, 6: 28, 10: 39,
    16: 53, 25: 70, 35: 86, 50: 105, 70: 134,
    95: 162, 120: 187, 150: 215, 185: 245, 240: 290
}

CALIBRES = [2, 4, 6, 10, 16, 20, 25, 32, 40, 50,
            63, 80, 100, 125, 160, 200, 250, 315, 400, 500, 630]

# ==========================================================
# 2. INTERFACE UTILISATEUR
# ==========================================================

style = {'description_width': 'initial'}
wide = widgets.Layout(width='350px')

titre = widgets.HTML(
    "<h2>⚡ SMART ELECTRICAL DESIGN</h2>"
    "<p>Prototype de calcul et de dimensionnement électrique BT</p>"
)

projet = widgets.Text(
    value="Projet électrique",
    description="Nom du projet :",
    style=style, layout=wide
)

depart = widgets.Text(
    value="Départ 01",
    description="Nom du départ :",
    style=style, layout=wide
)

reseau = widgets.Dropdown(
    options=["Triphasé 400 V", "Monophasé 230 V"],
    value="Triphasé 400 V",
    description="Réseau :",
    style=style, layout=wide
)

puissance = widgets.FloatText(
    value=15000,
    description="Puissance active (W) :",
    style=style, layout=wide
)

cosphi = widgets.FloatSlider(
    value=0.9, min=0.5, max=1.0, step=0.01,
    description="Cos φ :", readout_format='.2f',
    style=style, layout=wide
)

rendement = widgets.FloatSlider(
    value=1.0, min=0.5, max=1.0, step=0.01,
    description="Rendement η :", readout_format='.2f',
    style=style, layout=wide
)

longueur = widgets.FloatText(
    value=35,
    description="Longueur du câble (m) :",
    style=style, layout=wide
)

conducteur = widgets.Dropdown(
    options=["Cuivre", "Aluminium"],
    description="Conducteur :",
    style=style, layout=wide
)

pose = widgets.Dropdown(
    options=[
        "Pose en conduit / cheminement",
        "Pose sur chemin de câbles",
        "Pose enterrée"
    ],
    description="Mode de pose :",
    style=style, layout=wide
)

facteur_temp = widgets.FloatSlider(
    value=1.0, min=0.5, max=1.2, step=0.01,
    description="Facteur température :", readout_format='.2f',
    style=style, layout=wide
)

facteur_group = widgets.FloatSlider(
    value=1.0, min=0.5, max=1.2, step=0.01,
    description="Facteur groupement :", readout_format='.2f',
    style=style, layout=wide
)

reserve = widgets.FloatSlider(
    value=1.0, min=1.0, max=1.5, step=0.05,
    description="Facteur de réserve :", readout_format='.2f',
    style=style, layout=wide
)

source = widgets.Dropdown(
    options=["Transformateur HTA/BT", "Réseau BT"],
    description="Source :",
    style=style, layout=wide
)

puissance_tr = widgets.FloatText(
    value=250,
    description="Puissance transformateur (kVA) :",
    style=style, layout=wide
)

uk = widgets.FloatText(
    value=4,
    description="Tension de court-circuit Uk (%) :",
    style=style, layout=wide
)

icc_reseau = widgets.FloatText(
    value=10000,
    description="Icc réseau BT (A) :",
    style=style, layout=wide
)

protection_amont = widgets.IntText(
    value=250,
    description="Calibre protection amont (A) :",
    style=style, layout=wide
)

pouvoir_coupure = widgets.Dropdown(
    options=[4.5, 6, 10, 15, 25, 36, 50],
    value=10,
    description="Pouvoir de coupure Icu (kA) :",
    style=style, layout=wide
)

limite_du = widgets.FloatText(
    value=5,
    description="Limite ΔU (%) :",
    style=style, layout=wide
)

bouton = widgets.Button(
    description="CALCULER LE DIMENSIONNEMENT",
    button_style="success",
    icon="calculator",
    layout=widgets.Layout(width='350px', height='45px')
)

sortie = widgets.Output()

# ==========================================================
# 3. MOTEUR DE CALCUL
# ==========================================================

def calculer_dimensionnement(_):

    with sortie:
        clear_output()

        try:
            P = puissance.value * reserve.value
            L = longueur.value
            cos = cosphi.value
            eta = rendement.value

            if P <= 0 or L <= 0:
                raise ValueError("La puissance et la longueur doivent être positives.")

            if not 0 < cos <= 1 or not 0 < eta <= 1:
                raise ValueError("Vérifier le cos φ et le rendement.")

            if facteur_temp.value <= 0 or facteur_group.value <= 0:
                raise ValueError("Les facteurs de correction doivent être positifs.")

            mono = reseau.value == "Monophasé 230 V"
            U = 230 if mono else 400
            sinphi = math.sqrt(1 - cos**2)

            # Courant d'emploi
            if mono:
                Ib = P / (U * cos * eta)
            else:
                Ib = P / (math.sqrt(3) * U * cos * eta)

            S_app = P / (cos * eta) / 1000

            # Choix du conducteur
            rho = 0.0225 if conducteur.value == "Cuivre" else 0.036
            tableau = IZ_CU if conducteur.value == "Cuivre" else IZ_AL

            # Facteurs de correction
            K = facteur_temp.value * facteur_group.value

            # Recherche de section
            section_retenue = None
            Iz_retenu = None
            In_retenu = None

            for section in SECTIONS:
                Iz = tableau[section] * K

                calibre = next(
                    (c for c in CALIBRES if c >= Ib and c <= Iz),
                    None
                )

                if calibre is not None:
                    section_retenue = section
                    Iz_retenu = Iz
                    In_retenu = calibre
                    break

            if section_retenue is None:
                raise ValueError(
                    "Aucune section disponible dans la table indicative. "
                    "Augmenter la plage des sections ou vérifier les paramètres."
                )

            # Chute de tension
            # X = réactance linéique simplifiée, en ohm/km
            X_km = 0.08
            X = X_km / 1000

            if mono:
                delta_U = (
                    2 * L * Ib *
                    (rho * cos + X * 1000 * sinphi)
                    / section_retenue
                )
            else:
                delta_U = (
                    math.sqrt(3) * L * Ib *
                    (rho * cos + X * 1000 * sinphi)
                    / section_retenue
                )

            delta_U_pct = 100 * delta_U / U

            # ==================================================
            # COURANTS DE COURT-CIRCUIT APPROCHÉS
            # ==================================================

            if source.value == "Transformateur HTA/BT":
                Sn = puissance_tr.value * 1000
                uk_pct = uk.value

                if Sn <= 0 or uk_pct <= 0:
                    raise ValueError("Vérifier Sn et Uk du transformateur.")

                # Icc triphasé aux bornes BT du transformateur
                Icc1 = Sn / (math.sqrt(3) * U) * (100 / uk_pct)

            else:
                Icc1 = icc_reseau.value

            if Icc1 <= 0:
                raise ValueError("Le courant de court-circuit source doit être positif.")

            # Impédance équivalente de source par phase
            Zsource = U / (math.sqrt(3) * Icc1)

            # Impédance du câble par phase
            R_cable = rho * L / section_retenue
            X_cable = X * L
            Z_cable = math.sqrt(R_cable**2 + X_cable**2)

            # Estimation Icc triphasé à l'extrémité du câble
            Icc2 = U / (
                math.sqrt(3) * (Zsource + Z_cable)
            )

            # Estimation défaut phase-neutre en bout de câble
            Z_boucle = 2 * (Zsource + Z_cable)
            Icc3 = (U / math.sqrt(3)) / Z_boucle

            # ==================================================
            # PROTECTION
            # ==================================================

            Icu_A = pouvoir_coupure.value * 1000

            condition_ib = Ib <= In_retenu
            condition_iz = In_retenu <= Iz_retenu
            condition_icu = Icu_A >= Icc1

            # Sélectivité : impossible à confirmer sans les tableaux fabricant
            selectivite = "NON VÉRIFIÉE – tableaux fabricant nécessaires"

            # ==================================================
            # AFFICHAGE DES RÉSULTATS
            # ==================================================

            def statut(ok):
                return "✅ Conforme au critère calculé" if ok else "❌ Critère non satisfait"

            print("=" * 65)
            print("RÉSULTATS DU DIMENSIONNEMENT")
            print("=" * 65)

            print(f"Projet : {projet.value}")
            print(f"Départ : {depart.value}")
            print(f"Réseau : {reseau.value}")
            print(f"Puissance apparente : {S_app:.2f} kVA")
            print(f"Courant d'emploi Ib : {Ib:.2f} A")

            print("\n--- CÂBLE ---")
            print(f"Conducteur : {conducteur.value}")
            print(f"Mode de pose sélectionné : {pose.value}")
            print(f"Section proposée : {section_retenue} mm²")
            print(f"Courant admissible indicatif Iz : {Iz_retenu:.2f} A")
            print(f"Facteur global de correction : {K:.2f}")

            print("\n--- CHUTE DE TENSION ---")
            print(f"ΔU : {delta_U:.2f} V")
            print(f"ΔU (%) : {delta_U_pct:.2f} %")
            print(f"Limite définie : {limite_du.value:.2f} %")
            print(statut(delta_U_pct <= limite_du.value))

            print("\n--- COURANTS DE COURT-CIRCUIT ESTIMÉS ---")
            print(f"Icc1 : {Icc1 / 1000:.2f} kA – source BT")
            print(f"Icc2 : {Icc2 / 1000:.2f} kA – extrémité, défaut triphasé")
            print(f"Icc3 : {Icc3 / 1000:.2f} kA – extrémité, défaut phase-neutre")

            print("\n--- PROTECTION ---")
            print(f"Calibre proposé In : {In_retenu} A")
            print(f"Pouvoir de coupure sélectionné : {pouvoir_coupure.value} kA")
            print(f"Vérification Ib ≤ In : {statut(condition_ib)}")
            print(f"Vérification In ≤ Iz : {statut(condition_iz)}")
            print(f"Vérification Icu ≥ Icc source : {statut(condition_icu)}")

            print("\n--- SÉLECTIVITÉ ---")
            print(selectivite)

            print("\n" + "=" * 65)
            print("ATTENTION : résultats indicatifs, à valider avant toute utilisation.")
            print("Les calculs ne constituent pas une attestation de conformité.")
            print("=" * 65)

            # Tableau récapitulatif
            resultats = pd.DataFrame({
                "Paramètre": [
                    "Puissance apparente",
                    "Courant d'emploi Ib",
                    "Section proposée",
                    "Courant admissible Iz",
                    "Calibre In",
                    "Chute de tension",
                    "Icc1 source",
                    "Icc2 défaut triphasé en bout",
                    "Icc3 défaut phase-neutre en bout"
                ],
                "Résultat": [
                    f"{S_app:.2f} kVA",
                    f"{Ib:.2f} A",
                    f"{section_retenue} mm²",
                    f"{Iz_retenu:.2f} A",
                    f"{In_retenu} A",
                    f"{delta_U_pct:.2f} %",
                    f"{Icc1/1000:.2f} kA",
                    f"{Icc2/1000:.2f} kA",
                    f"{Icc3/1000:.2f} kA"
                ]
            })

            display(resultats)

        except Exception as e:
            print("Erreur :", str(e))


bouton.on_click(calculer_dimensionnement)

# ==========================================================
# 4. ORGANISATION DU DASHBOARD
# ==========================================================

onglets = widgets.Tab(children=[
    widgets.VBox([projet, depart, reseau, puissance, cosphi, rendement]),
    widgets.VBox([longueur, conducteur, pose, facteur_temp,
                  facteur_group, reserve, limite_du]),
    widgets.VBox([source, puissance_tr, uk, icc_reseau]),
    widgets.VBox([protection_amont, pouvoir_coupure])
])

onglets.set_title(0, "Projet et réseau")
onglets.set_title(1, "Câbles")
onglets.set_title(2, "Source et Icc")
onglets.set_title(3, "Protections")

display(titre)
display(onglets)
display(bouton)
display(sortie)
```
